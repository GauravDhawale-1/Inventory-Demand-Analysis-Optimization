```python
import pandas as pd
import numpy as np

df= pd.read_csv("Inventory Project Datasset.csv")


print(df.info())
print(df.isnull().sum())
print(df.describe(include="all"))
```

```python
print("unique Product:",df["Product_Code"].nunique())
print("Warehouse:",df["Warehouse"].nunique())
print("Product Category:",df["Product_Category"].nunique())
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df["Date"]=pd.to_datetime(
    df["Date"],
    errors="coerce"
)

print("Missing Date :", df["Date"].isna().sum())
print("Date Range :",df["Date"].min(),df["Date"].max())
```

```python
demand = df["Order_Demand"].astype("string").str.strip()

is_negative = demand.str.match(r"^\(.*\)$", na=False)

demand = demand.str.replace(r"[(),\s]", "", regex=True)

df["Order_Demand"] = pd.to_numeric(
    demand.where(~is_negative, "-" + demand),
    errors="coerce"
)

print(df["Order_Demand"].describe())
print("Invalid quantities:", df["Order_Demand"].isna().sum())
```

```python
print("Exacat Duplicate :" , df.duplicated().sum())

duplicates=df[df.duplicated(keep=False)]
print(duplicates.head(30))

df=df.drop_duplicates().copy()
```

```python
print("Missing Products Code :", df["Product_Code"].isna().sum())
print("Missing werehouse :", df["Warehouse"].isna().sum())
print("Missing Category :", df["Product_Category"].isna().sum())
print("Missing Order_Demands:",df["Order_Demand"].isna().sum())

print("Negative Demands Rows :", (df["Order_Demand"] < 0 ).sum())
print("Zero Demand Rows :",(df["Order_Demand"]==0).sum())
```

```python
requried=["Product_Code","Warehouse","Product_Category","Date","Order_Demand"]
clean_df= df.dropna(subset=requried).copy()

print("Row Before:",len(df))
print("Row After:",len(clean_df))
print("Row Diffrence:", len(df) - len(clean_df))
```

```python
print("Final Shape:",clean_df.shape)
print("Datatypes :",clean_df.dtypes)
print("Missing Values :",clean_df.isnull().sum())
print("Exacet Duplicate :",clean_df.duplicated().sum())
```

```python
clean_df.to_csv("Cleaned Inventory Project Datasset.csv", index=False)
print("Cleaned Inventory Project Datasset Seved Successfully")
```
